1. Escriba una función que se llame `mybin` que reciba un número entero real de 16 bits y realice su conversión en código binario, compare su resultado con `bin(4321)`

````python
binario =[]

def mybinario(x):
  while x > 0:
    if x % 2 == 0:
      binario.append(0)
    elif x % 2 != 0:
     binario.append(1)
    x//= 2
  return binario[::-1]

mybinario(4321)
````

````python
bin(4321)
````

2. Cree una función llamada `myint` pero que a partir de un número binario con base de 16 bits encuentre el entero correspondiente. Compare su resultado con `int(0b1000011100001)`

````python
def myentero(x):
  x = [int(digito) for digito in str(x)]
  x = x[::-1]
  entero = 0
  for n in range(len(x)):
    if x[n] == 1:
      entero += 2**n
  return entero

print(myentero(1000011100001))
print(int(0b1000011100001))
````

3. Generar una función llamada `number64` para flotantes de precisión doble (64bits) donde a partir de un número binario encuentre el valor real. Compruebe su solución usando

```
number64("0100000000111011100100001111111111111111111111111111111111111111")
```



````python
def myenteroneg(x):
  x = [int(digito) for digito in str(x)]
  entero = 0
  for n in range(len(x)):
    if x[n] == 1:
      entero += 2**(-(n+1))
  return entero
````

````python
def numero64(y):
  y = [int(digito) for digito in str(y)]
  exponente = y[1:12]
  mantiza = y[12:64]
  exponente_n = int("".join(map(str,exponente)),2)
  E = exponente_n - 1023
  mantiza_n = int("".join(map(str,mantiza)))
  mantiza_fraccionaria = myenteroneg(mantiza_n)
  signo = int(y[0])
  valor = ((-1)**signo)*(1 + mantiza_fraccionaria ) * (2**(E))
  return valor

numero64("0100000000111011100100001111111111111111111111111111111111111111")
````
